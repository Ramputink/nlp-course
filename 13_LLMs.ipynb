# 1 .- Introducción a los modelos de lenguaje (LMs)

Un **modelo de lenguaje** es un **sistema diseñado para comprender, generar y manipular el lenguaje humano**.

Su función principal es **analizar una secuencia de palabras de entrada ("query")** y predecir cuál es la palabra (o "token") que tiene más probabilidades de ir a continuación, basándose en el contexto, a fin de **producir la respuesta**.

Los conceptos clave que definen cómo funcionan estos modelos son:

**1. La Arquitectura Base: El Transformer**

La gran mayoría de los modelos de lenguaje modernos (como Gemini, GPT o Llama) se basan en la **arquitectura de red neuronal llamada Transformer y el mecanismo de atención visto en el tema anterior**. Esto permite al modelo **"prestar atención" a todas las palabras del contexto al mismo tiempo** (en lugar de procesarlas secuencialmente), lo que le ayuda a comprender el contexto, los matices y las relaciones a largo plazo entre las palabras, incluso en textos muy largos.

Casi todos los LLMs más avanzados actuales —incluyendo la familia **GPT, Claude, Grok, Llama y Gemini**— se basan fundamentalmente en una **arquitectura Transformer solo de decodificador** ("decoder-only"). Aunque el artículo original sobre Transformers ("Attention Is All You Need") incluía tanto un codificador (para procesar la entrada) como un decodificador (para generar la salida), la industria de la IA se dio cuenta de que, para la generación masiva de lenguaje, bastaba solo con el decodificador.

**2. Aprendizaje del lenguaje con datos masivos**

Durante esta fase de **entrenamiento previo (Pre-training)**, **se alimenta al modelo con grandes cantidades de texto** proveniente de internet, libros, artículos y código fuente. Durante este proceso, el modelo intenta predecir palabras ocultas en esos textos. A base de ensayo y error (**ajustando millones o miles de millones de parámetros**), el modelo internaliza la gramática, los hechos sobre el mundo, el razonamiento lógico y hasta el humor.

**3. Ajuste Fino (Fine-Tuning)**

Un modelo recién entrenado solo sabe predecir texto, pero no sabe cómo comportarse como un asistente útil. Por eso, pasa por una fase de ajuste fino. Aquí se le **entrena con ejemplos específicos de preguntas y respuestas** para que aprenda a seguir instrucciones, ser educado, responder en formatos específicos y evitar generar contenido tóxico o peligroso.

**4. Capacidades Principales**

Gracias a este proceso, los Modelos de Lenguaje (LLMs) pueden realizar **tareas sorprendentes sin haber sido programados explícitamente para cada una de ellas**:

- **Generación de texto**: Escribir ensayos, poemas, correos electrónicos o código de programación.

- **Comprensión y resumen**: Extraer las ideas principales de documentos largos.

- **Traducción**: Traducir textos entre decenas de idiomas con alta precisión contextual.

- **Razonamiento y análisis**: Responder preguntas complejas y resolver problemas lógicos o matemáticos.



Existe infinidad de modelos de IA, tanto para su uso local como para su uso comercial. Una de las características más distintiva es el **tamaño del modelo**.

El **número de parámetros en un modelo de lenguaje** representa el **número de "conexiones neuronales" o "pesos" que posee**. A mayor cantidad de parámetros, el modelo suele tener mayor capacidad de razonamiento, memoria y comprensión de matices, pero también requiere mucha más potencia computacional (hardware) para funcionar.

En la industria de la IA, se usa la letra "**B**" para referir a **Billions** (mil millones en español) y "**T**" para **Trillions** (billones en español).

Un posible taxonomía estructurada de modelos atendiendo a su tamaño se indica en las siguientes secciones.


#  2.-Modelos Nano y Micro (Menos de 1B de parámetros)

Son la **categoría más pequeña y ágil**. Están diseñados para tareas muy específicas y sencillas donde la velocidad y la eficiencia energética son críticas. Pueden **ejecutarse en dispositivos móviles (smartphones) o dispositivos de la "Internet de las Cosas" (IoT)** sin necesidad de conexión a internet. Consumen muy poca memoria RAM. Sus posibles casos de uso son: Autocompletado básico de texto en teclados móviles, clasificación simple de sentimientos o tareas básicas de domótica.

**Ejemplos**: `MobileBERT, ALBERT, TinyBERT`.

# 3.-Modelos Pequeños o SLMs (Small Language Models) (1B a 10B)

Los **SLMs** son actualmente una de las áreas de mayor innovación porque ofrecen un **equilibrio excelente entre rendimiento y accesibilidad**. Se pueden ejecutar de forma local en **ordenadores portátiles de consumo (laptops) o servidores económicos**. Son muy rápidos y baratos de operar. Posibles casos de uso: Asistentes locales, resumen de documentos de uso diario, generación de código ligero, o ajuste fino (fine-tuning) para herramientas corporativas internas y privadas.

**Ejemplos**: `Phi-3 Mini (3.8B), Mistral (7B), Gemma (2B y 7B), Llama 3 (8B)`.

# 4.- Modelos Medianos (10B a 50B)

Entramos en la categoría donde los modelos empiezan a mostrar capacidades avanzadas de razonamiento lógico, aunque ya **requieren hardware especializado para funcionar con fluidez**. Normalmente necesitan **una o varias tarjetas gráficas (GPUs) de gama alta** para ejecutarse localmente de forma eficiente, o bien servidores dedicados en la nube. Como casos de uso: Sistemas de atención al cliente automatizados de alta calidad, asistentes de programación robustos, redacción de contenido profesional y análisis de datos complejos.

**Ejemplos**: `Mixtral 8x7B (un modelo que combina varios más pequeños), Cohere Command R (35B), Llama 2 (13B)`.

# 5.-Grandes Modelos o LLMs (Large Language Models) (50B a 300B)

Estos son los **"pesos pesados" estándar de la industria**. Poseen un **amplio conocimiento del mundo y excelentes capacidades multilingües y lógicas**. Imposibles de ejecutar en un ordenador personal convencional. Requieren **clústeres de servidores en centros de datos con decenas de GPUs empresariales trabajando en paralelo**.

Casos de uso: Investigación científica, resolución de problemas matemáticos complejos, traducción de alto nivel, asistentes virtuales de uso general y toma de decisiones autónoma en agentes de IA.

**Ejemplos**: `Llama 3 (70B), Falcon (180B), Grok-1 (314B)`.



# 6.-Modelos Masivos o de Frontera (Frontier Models) (Más de 300B a 1T+)

Son los **modelos más potentes del planeta**, entrenados por las empresas tecnológicas más grandes. Muchos de ellos utilizan una arquitectura llamada "Mezcla de Expertos" (MoE), donde el modelo está dividido en sub-redes para ser más eficiente. Su tamaño exacto suele ser un secreto comercial. **Requieren superordenadores valorados en cientos de millones de dólares solo para su entrenamiento y mantenimiento**. Representan el estado del arte (SOTA) en inteligencia artificial. Como casos de uso: Sistemas multimodales avanzados (que entienden texto, imagen, audio y video de forma nativa simultáneamente), programación de software de nivel experto, razonamiento profundo y prolongado.

**Ejemplos (2026)**: `GPT-5.2 (3.5-5T), Grok 4.1 (1.3-2T), Claude 4.5 (1.5-3T), Gemini 3.1 (2-3T), Llama 4 (2T)`.

Hoy en día, el "número de parámetros" ya no es la única métrica que importa. La eficiencia con la que estos modelos usan herramientas externas, el tamaño de su ventana de contexto (cuántos "tokens" pueden leer en un solo mensaje) y sus capacidades de razonamiento interno se han vuelto factores mucho más decisivos que el simple tamaño bruto.

## Ejemplo de uso de modelo micro: SmolLM2

Contiene **360M** parámetros.

Enlace de búsqueda en Hugging Face: ``HuggingFaceTB/SmolLM2-360M-Instruct``


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
transformers.logging.set_verbosity_error()

model_id = "HuggingFaceTB/SmolLM2-360M-Instruct" # respuestas pobres

print("Cargando el modelo {}...".format(model_id))

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto" # Automatically maps the model to available devices (CPU/GPU)
)

# Configure the pipeline for text generation with the SmolLM2 model
generator_smol = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    device_map="auto"
)

print("Modelo {} cargado exitosamente.".format(model_id))
```

```python
def AI(prompt):
  output_smol = generator_smol(
    prompt,
    max_new_tokens=256,  # Limit the length of the generated text
    num_return_sequences=1,
    do_sample=True,     # Enable sampling for more creative output
    temperature=0.7,    # Control the randomness of the generation
    top_k=50,           # Consider only the top 50 most likely tokens
    top_p=0.95,          # Consider tokens whose cumulative probability is less than 0.95
    max_length=None
  )

  print(textwrap.fill(output_smol[0]["generated_text"], width=80))
```

```python
AI("Hola!")
```

### Chat libre: ejecuta las siguientes celdas para entrar en modo "chat".

```python
MAX_INTERACCIONES = 5  # El bot recordará tus últimas 5 preguntas (y sus 5 respuestas)
MAX_MENSAJES = (MAX_INTERACCIONES * 2) + 1  # Multiplicamos por 2 (pregunta+respuesta) y sumamos 1 (el prompt del sistema)
historial_chat = [
    {"role": "system", "content": "Eres un asistente útil, amigable y que responde de forma concisa en español."}
]

def procesa_query(entrada_usuario):
  historial_chat.append({"role": "user", "content": entrada_usuario})
  resultado = generator_smol(historial_chat, max_new_tokens=256, max_length=None, return_full_text=False, temperature=0.7)
  respuesta_modelo = resultado[0]['generated_text']
  print(f"\nAsistente: {respuesta_modelo}\n" + "-" * 50)
  historial_chat.append({"role": "assistant", "content": respuesta_modelo})
  if len(historial_chat) > MAX_MENSAJES:
    historial_chat.pop(1) # Borramos la pregunta más antigua
    historial_chat.pop(1) # Borramos la respuesta más antigua (que ahora pasó a ocupar el índice 1)
```

```python
print("=" * 50 + "\n🤖 Chatbot iniciado, escribe 'salir' para terminar la conversación.\n" + "=" * 50 + "\n")
while True:
  entrada_usuario = input("Tú: ")
  if entrada_usuario.lower() in ["salir", "exit", "quit", "terminar"]:
    print("\n¡Chat terminado! Hasta la próxima.")
    break
  procesa_query(entrada_usuario)
```

# Ejecución local más eficiente de modelos LLM de mayor tamaño

Para ejecutar en local modelos mayores es conveniente usar un formato de archivo especial para la descarga del modelo.

**1. El formato GGUF y el "Offloading"**

**GGUF** (GPT-Generated Unified Format) es un formato de archivo diseñado específicamente para ejecutar modelos de lenguaje (LLMs) de manera ultraeficiente **en hardware de consumo, especialmente en CPUs y sistemas con memoria limitada**.

En lugar de descargar los parámetros ("tensores") puros del modelo, **descargamos una versión del modelo empaquetada en formato GGUF**. Este formato está diseñado para permitir GPU Offloading (Descarga a la GPU): el modelo se carga principalmente en la memoria RAM normal del ordenador (que es mucho más barata y abundante que la memoria de la tarjeta gráfica). El motor de ejecución envía a la tarjeta gráfica solo las capas matemáticas que caben en ella para acelerar el proceso, y el resto lo calcula el procesador (CPU). Como resultado **se puede ejecutar un modelo masivo, aunque más lento, aunque sea en una CPU en vez de GPU**.



Existen **herramientas para ejecutar modelos GGUF**, son motores de ejecución optimizados en C++, muy fáciles de usar.

- La **opción para terminal** es **Ollama**: Es la herramienta más popular para desarrolladores. Tras instalarla se abre una consola de comandos y se puede acceder a los modelos con órdenes del tipo: `ollama run llama3.1:70b`. Descarga el modelo cuantizado, analiza el hardware disponible y lo divide entre RAM y VRAM automáticamente.

- **La opción para uso desde python es Llama.cpp**: Es el **motor puro de código abierto que está debajo de Ollama y LM Studio**. Se usa desde la terminal o desde Python y da control absoluto sobre cuántos hilos del procesador usar y cómo distribuir la memoria.

- **LM Studio** es la **opción con interfaz gráfica**, un programa de escritorio que permite buscar cualquier modelo en Hugging Face directamente desde su buscador (buscando siempre la palabra "GGUF"). Tiene un panel lateral con un deslizador llamado GPU Offload.

Veamos a continuación ejemplos de uso de `Llama.cpp` desde Python y de `LM studio`.

## Ejemplo de uso de modelo SLM Phi-3.5-mini-instruct desde Python usando Llama.cpp

Contiene **3.8B** parámetros.

Enlace de búsqueda en Hugging Face: ``HuggingFaceTB/microsoft/Phi-3.5-mini-instruct``


```python
# Instalación de llama-cpp-python con aceleración Metal en Apple Silicon (M2 Pro).
# Se compila desde fuente activando el backend Metal para usar la GPU integrada.
import sys, subprocess, platform

# Si ya está instalado, lo ignoramos.
try:
    import llama_cpp
    print(f"llama-cpp-python ya instalado: {llama_cpp.__version__}")
except ImportError:
    print("Instalando llama-cpp-python con soporte Metal...")
    env_flags = []
    if platform.system() == "Darwin":
        # Metal = GPU integrada del M2 Pro. Mucho más rápido que CPU.
        env_flags = ["CMAKE_ARGS=-DGGML_METAL=on"]

    cmd = env_flags + [
        sys.executable, "-m", "pip", "install", "--upgrade", "--force-reinstall",
        "--no-cache-dir", "llama-cpp-python",
    ]
    # En macOS pip respeta CMAKE_ARGS si lo pasamos por env, no por argv.
    import os
    env = os.environ.copy()
    if platform.system() == "Darwin":
        env["CMAKE_ARGS"] = "-DGGML_METAL=on"
        env["FORCE_CMAKE"] = "1"
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "--upgrade", "--force-reinstall",
         "--no-cache-dir", "llama-cpp-python"],
        env=env,
    )

    # Re-importar tras la instalación
    import importlib, llama_cpp
    importlib.reload(llama_cpp)
    print(f"Instalado: llama-cpp-python {llama_cpp.__version__}")

# Para que detecte huggingface y descargue el modelo automáticamente
try:
    import huggingface_hub  # noqa: F401
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "huggingface_hub"])
```

```python
from llama_cpp import Llama
import multiprocessing as mp

# Optimizaciones para M2 Pro (10-12 cores, GPU integrada con Metal):
#  - n_threads: hilos de la CPU para las capas que NO se descargan a GPU
#  - n_gpu_layers=-1: descarga TODAS las capas a la GPU Metal (máxima velocidad)
#  - n_ctx: ventana de contexto. 4096 cabe sin problema en 16 GB.

llm = Llama.from_pretrained(
    repo_id="bartowski/Phi-3.5-mini-instruct-GGUF",
    filename="Phi-3.5-mini-instruct-Q4_K_M.gguf",
    n_threads=max(1, mp.cpu_count() - 2),
    n_gpu_layers=-1,
    n_ctx=4096,
    n_batch=512,
    verbose=False,
)
print("Modelo Phi-3.5-mini-instruct cargado con Metal.")
```

```python
def AI(prompt):
  resultado = llm.create_chat_completion(
      messages = [
      {"role": "system", "content": "Eres un asistente útil y directo."},
      {"role": "user", "content": prompt}
    ]
  )

  print(textwrap.fill(resultado["choices"][0]["message"]["content"], width=80))
```
